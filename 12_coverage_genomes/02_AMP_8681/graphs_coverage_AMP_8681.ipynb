```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Set seaborn style
sns.set_style("white")

# Function to plot the coverage per base of a genome with and without zoom
def coverage_graph(input, output, amp_start, amp_end, zoom_start=None, zoom_end=None, title=None, total_coverage=None, color='blue', y_label='RNA-seq depth'):
    df = pd.read_csv(input, sep='\t', names=['genome','base', 'coverage'])

    # Create the plot using seaborn
    plt.figure(figsize=(7, 5))
    sns.lineplot(data=df, x='base', y='coverage', color=color, linewidth=1.5)
    plt.title(title + '\n' + 'Total coverage ' + total_coverage, fontsize=14, pad=20)
    plt.xlabel('Genome position', fontsize=12)
    plt.ylabel(y_label, fontsize=12)
    plt.axvline(x=amp_start, color='red', linestyle='--', alpha=0.8, linewidth=2)
    plt.axvline(x=amp_end, color='red', linestyle='--', alpha=0.8, linewidth=2)
    sns.despine()
    plt.tight_layout()
    plt.savefig(output, dpi=300, bbox_inches='tight')
    plt.savefig(output.replace('.png', '.svg'), bbox_inches='tight')
    plt.clf()

    if zoom_start and zoom_end:
        plt.figure(figsize=(7, 5))
        sns.lineplot(data=df, x='base', y='coverage', color=color, linewidth=1.5)
        plt.xlim(zoom_start, zoom_end)
        plt.title(title + '\n' + 'Total coverage: ' + total_coverage, fontsize=14, pad=20)
        plt.xlabel('Genome position', fontsize=12)
        plt.ylabel(y_label, fontsize=12)
        plt.axvline(x=amp_start, color='red', linestyle='--', alpha=0.8, linewidth=2)
        plt.axvline(x=amp_end, color='red', linestyle='--', alpha=0.8, linewidth=2)
        sns.despine()
        plt.tight_layout()
        plt.savefig(output.replace('.png', '_zoom.png'), dpi=300, bbox_inches='tight')
        plt.savefig(output.replace('.png', '_zoom.svg'), bbox_inches='tight')
        plt.clf()
        
        
# Execution of the function for each genome
coverage_graph(input='./BK038133.1_allsamples_aligned_genomecov.txt',
               output='./BK038133.1_allsamples_aligned_genomecov.png',
               amp_start=116,
               amp_end=463,
            #    zoom_start=2467298,
            #    zoom_end=2487761,
               title='RNA-seq Coverage across Caudoviricetes sp. isolate ctlNO7 genome',
               total_coverage='12.82%',
               color='turquoise',
               y_label='RNA-seq depth')

coverage_graph(input='./CP107216.1_allsamples_aligned_genomecov.txt',
               output='./CP107216.1_allsamples_aligned_genomecov.png',
               amp_start=2213911,
               amp_end=2214257,
            #    zoom_start=2467298,
            #    zoom_end=2487761,
               title='RNA-seq Coverage across Clostriales bacterium isolate KR001 HIC 0007 genome',
               total_coverage='39.83%',
               color='orange',
               y_label='RNA-seq depth')

coverage_graph(input='./01_besthit_Caudoviricetes_sp._isolate_ctlNO7_viral_allsamples_aligned_genomecov.txt',
               output='./01_besthit_Caudoviricetes_sp._isolate_ctlNO7_viral_allsamples_aligned_genomecov.png',
               amp_start=116,
               amp_end=463,
            #    zoom_start=2467298,
            #    zoom_end=2487761,
               title='Phageome Coverage across Caudoviricetes sp. isolate ctlNO7 genome',
               total_coverage='7.18%',
               color='lightgreen',
               y_label='Phageome DNA reads depth')
```
