```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Set seaborn style
sns.set_style("white")

# Function to plot the coverage per base of a genome with and without zoom
def coverage_graph(input, output, amp_start, amp_end, zoom_start=None, zoom_end=None, title=None, total_coverage=None, color='blue', y_label='RNA-seq depth'):
    df = pd.read_csv(input, sep='\t', names=['genome','base', 'coverage'])

    # Create the plot using seaborn
    plt.figure(figsize=(7, 5))
    sns.lineplot(data=df, x='base', y='coverage', color=color, linewidth=1.5)
    plt.title(title + '\n' + 'Total coverage ' + total_coverage, fontsize=14, pad=20)
    plt.xlabel('Genome position', fontsize=12)
    plt.ylabel(y_label, fontsize=12)
    plt.axvline(x=amp_start, color='red', linestyle='--', alpha=0.8, linewidth=2)
    plt.axvline(x=amp_end, color='red', linestyle='--', alpha=0.8, linewidth=2)
    sns.despine()
    plt.tight_layout()
    plt.savefig(output, dpi=300, bbox_inches='tight')
    plt.savefig(output.replace('.png', '.svg'), bbox_inches='tight')
    plt.clf()

    if zoom_start and zoom_end:
        plt.figure(figsize=(7, 5))
        sns.lineplot(data=df, x='base', y='coverage', color=color, linewidth=1.5)
        plt.xlim(zoom_start, zoom_end)
        plt.title(title + '\n' + 'Total coverage: ' + total_coverage, fontsize=14, pad=20)
        plt.xlabel('Genome position', fontsize=12)
        plt.ylabel(y_label, fontsize=12)
        plt.axvline(x=amp_start, color='red', linestyle='--', alpha=0.8, linewidth=2)
        plt.axvline(x=amp_end, color='red', linestyle='--', alpha=0.8, linewidth=2)
        sns.despine()
        plt.tight_layout()
        plt.savefig(output.replace('.png', '_zoom.png'), dpi=300, bbox_inches='tight')
        plt.savefig(output.replace('.png', '_zoom.svg'), bbox_inches='tight')
        plt.clf()
        
        
# Execution of the function for each genome
coverage_graph(input='./BK026787.1_allsamples_aligned_genomecov.txt',
               output='./BK026787.1_allsamples_aligned_genomecov.png',
               amp_start=19543,
               amp_end=19898,
            #    zoom_start=9526,
            #    zoom_end=29989,
               title='RNA-seq Coverage across Caudorviricetes sp. isolate ctJ1L4 genome',
               total_coverage='14.06%',
               color='turquoise')

coverage_graph(input='./CP133078.1_allsamples_aligned_genomecov.txt',
               output='./CP133078.1_allsamples_aligned_genomecov.png',
               amp_start=2477315,
               amp_end=2477669,
            #    zoom_start=2467298,
            #    zoom_end=2487761,
               title='RNA-seq Coverage across Anaerotignum sp. MB30-C6 genome',
               total_coverage='0.67%',
               color='orange')

coverage_graph(input='./01_besthit_Caudorviricetes_sp._isolate_ctJ1L4_viral_allsamples_aligned_genomecov.txt',
               output='./01_besthit_Caudorviricetes_sp._isolate_ctJ1L4_viral_allsamples_aligned_genomecov.png',
               amp_start=19543,
               amp_end=19898,
               # zoom_start=18543,
               # zoom_end=20898,
               title='Phageome Coverage across Caudorviricetes sp. isolate ctJ1L4 genome',
               total_coverage='70.46%',
               color='lightgreen',
               y_label='Phageome DNA reads depth')
```
