```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Set seaborn style
sns.set_style("white")

# Function to plot the coverage per base of a genome with and without zoom
def coverage_graph(input, output, amp_start, amp_end, zoom_start=None, zoom_end=None, title=None, total_coverage=None, color='blue'):
    df = pd.read_csv(input, sep='\t', names=['genome','base', 'coverage'])

    # Create the plot using seaborn
    plt.figure(figsize=(7, 5))
    sns.lineplot(data=df, x='base', y='coverage', color=color, linewidth=1.5)
    plt.title(title + '\n' + 'Total coverage ' + total_coverage, fontsize=14, pad=20)
    plt.xlabel('Genome position', fontsize=12)
    plt.ylabel('RNA-seq depth', fontsize=12)
    plt.axvline(x=amp_start, color='red', linestyle='--', alpha=0.8, linewidth=2)
    plt.axvline(x=amp_end, color='red', linestyle='--', alpha=0.8, linewidth=2)
    sns.despine()
    plt.tight_layout()
    plt.savefig(output, dpi=300, bbox_inches='tight')
    plt.savefig(output.replace('.png', '.svg'), bbox_inches='tight')
    plt.clf()

    if zoom_start and zoom_end:
        plt.figure(figsize=(7, 5))
        sns.lineplot(data=df, x='base', y='coverage', color=color, linewidth=1.5)
        plt.xlim(zoom_start, zoom_end)
        plt.title(title + '\n' + 'Total coverage: ' + total_coverage, fontsize=14, pad=20)
        plt.xlabel('Genome position', fontsize=12)
        plt.ylabel('RNA-seq depth', fontsize=12)
        plt.axvline(x=amp_start, color='red', linestyle='--', alpha=0.8, linewidth=2)
        plt.axvline(x=amp_end, color='red', linestyle='--', alpha=0.8, linewidth=2)
        sns.despine()
        plt.tight_layout()
        plt.savefig(output.replace('.png', '_zoom.png'), dpi=300, bbox_inches='tight')
        plt.savefig(output.replace('.png', '_zoom.svg'), bbox_inches='tight')
        plt.clf()
        
# Execution of the function for each genome
coverage_graph(input='./CP126057.1_allsamples_aligned_genomecov.txt',
               output='./CP126057.1_allsamples_aligned_genomecov.png',
               amp_start=71409,
               amp_end=72152,
            #    zoom_start=9526,
            #    zoom_end=29989,
               title='RNA-seq Coverage across P. dorei plasmid genome',
               total_coverage='95.55%',
               color='turquoise')

coverage_graph(input='./CP046424.1_allsamples_aligned_genomecov.txt',
               output='./CP046424.1_allsamples_aligned_genomecov.png',
               amp_start=1750994,
               amp_end=1751737,
            #    zoom_start=2467298,
            #    zoom_end=2487761,
               title='RNA-seq Coverage across Phocaeicola dorei genome',
               total_coverage='74.73%',
               color='orange')

```
